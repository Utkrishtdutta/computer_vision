```python
from deepface import DeepFace
import cv2
import matplotlib.pyplot as plt
```

```python
img_path = 'wedding_pic.jpg'
img = cv2.imread(img_path)
plt.imshow(img[:,:,::-1])
```

## Face Detection

```python
face = DeepFace.extract_faces(img_path,
                           detector_backend = 'yolov8',
                           align=True)

len(face)
```

```python
plt.imshow(face[0]['face'])
```

```python
plt.imshow(face[1]['face'])
```

```python
plt.imshow(face[2]['face'])
```

## Image Analyst

```python
obj = DeepFace.analyze(img_path,
                           detector_backend = 'yolov8',
                           align=True)
```

```python
for i in range(len(obj)):
    print(f"{i}) ", obj[i]['dominant_emotion'], obj[i]['dominant_gender'], obj[i]['age'])
```

## Verify

```python
res = DeepFace.verify(img1_path = "sachin1.jpg", 
                      img2_path = "sachin2.jpg",
                      detector_backend = 'yolov8')

res['verified']
```

## Embedding

```python
embedding_res = DeepFace.represent(
  img_path = "sachin1.jpg", 
  detector_backend = 'yolov8',
  align = True,
)

embedding_res[0]['embedding']
```
